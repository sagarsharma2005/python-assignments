```python
#1) Create a CSV file for address book, CSV file should have column for name, address, mobile, email. Insert 2-3 dummy data entered by user.

import csv

file = open("address_book.csv", "w", newline="")

writer = csv.writer(file)

writer.writerow(["Name", "Address", "Mobile", "Email"])

for i in range(3):
    name = input("Enter Name: ")
    address = input("Enter Address: ")
    mobile = input("Enter Mobile: ")
    email = input("Enter Email: ")

    writer.writerow([name, address, mobile, email])

file.close()

print("CSV file created successfully")
```

```python
#2) Create 2-3 tables

import sqlite3

conn = sqlite3.connect("student.db")

cur = conn.cursor()

cur.execute("drop table if exists student")

cur.execute("drop table if exists teacher")

cur.execute("drop table if exists marks")

cur.execute("create table if not exists student(id integer, name text, class text)")

cur.execute("create table if not exists teacher(id integer, name text, subject text)")

cur.execute("create table if not exists marks(id integer, student_id integer, marks integer)")

conn.commit()

print("Tables created successfully")

conn.close()
```

```python
#3) Insert some records

import sqlite3

conn = sqlite3.connect("student.db")

cur = conn.cursor()

cur.execute("insert into student values(1,'Sagar','BTech')")
cur.execute("insert into student values(2,'Rahul','BCA')")

cur.execute("insert into teacher values(1,'Amit','Python')")
cur.execute("insert into teacher values(2,'Neha','DBMS')")

cur.execute("insert into marks values(1,1,85)")
cur.execute("insert into marks values(2,2,90)")

conn.commit()

print("Records inserted successfully")

conn.close()
```

```python
#4) Perform diffrent select operations

import sqlite3

conn = sqlite3.connect("student.db")

cur = conn.cursor()

print("Student Table")
for i in cur.execute("select * from student"):
    print(i)

print("Teacher Table")
for i in cur.execute("select * from teacher"):
    print(i)

print("Marks Greater Than 85")
for i in cur.execute("select * from marks where marks > 85"):
    print(i)

print("Only Student Names")
for i in cur.execute("select name from student"):
    print(i)

conn.close()
```

```python
#5) Update some data

import sqlite3

conn = sqlite3.connect("student.db")

cur = conn.cursor()

cur.execute("update student set name='Sagar Sharma' where id=1")

cur.execute("update marks set marks=95 where id=1")

conn.commit()

print("Data updated successfully")

conn.close()
```

```python
#6) Delete some data


import sqlite3

conn = sqlite3.connect("student.db")

cur = conn.cursor()

cur.execute("delete from student where id=2")

cur.execute("delete from marks where id=2")

conn.commit()

print("Data deleted successfully")

conn.close()
```
