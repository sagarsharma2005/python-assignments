```python
#1) Practise Pandas Series
	#Create a Pandas Series from Dictionary 
	

import pandas as pd

data = {
    "a": 10,
    "b": 20,
    "c": 30
}

s = pd.Series(data)

print(s)

#Create a Pandas Series from Lists 

import pandas as pd

data = [10, 20, 30, 40, 50]

s = pd.Series(data)

print(s)

#Access the elements of a Series in Pandas

import pandas as pd

data = [100, 200, 300, 400]

s = pd.Series(data)

print(s[0])
print(s[2])







```

```python
#2) DataFrames
#   Make a Pandas DataFrame with a two-dimensional Python list 

import pandas as pd

data = [
    [1, "Sagar", 85],
    [2, "Rahul", 90],
    [3, "Aman", 78]
]

df = pd.DataFrame(data)

print(df)

#Create DataFrame from Python dict 

import pandas as pd

data = {
    "Name": ["Sagar", "Rahul", "Aman"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)

#Create Pandas dataframe using list of lists

import pandas as pd

data = [
    ["Sagar", 85],
    ["Rahul", 90],
    ["Aman", 78]
]

df = pd.DataFrame(data, columns=["Name", "Marks"])

print(df)

#Create a Pandas dataframe using list of tuples 

import pandas as pd

data = [
    ("Sagar", 85),
    ("Rahul", 90),
    ("Aman", 78)
]

df = pd.DataFrame(data, columns=["Name", "Marks"])

print(df)

#Create a Pandas DataFrame from List of Dicts

import pandas as pd

data = [
    {"Name": "Sagar", "Marks": 85},
    {"Name": "Rahul", "Marks": 90},
    {"Name": "Aman", "Marks": 78}
]

df = pd.DataFrame(data)

print(df)







```
